## PyTorch is a Machine learning framework where AI models can be implemented and saved for inference.

#### It can be used by importing torch

```python
import torch
```

#### PyTorch works with tensors. A tensor is a multidimensional array storing the elements with same datatype. To convert to tensor, tensor() function is used.


#### i) Scalar is a 0D array. It stores just the number (the magnitude).
#### To find number of dimensions, ndim attribute is used.

```python
import torch

scalar = torch.tensor(1000000)
print(scalar)
print(scalar.ndim)
```

#### ii) Vector is a 1D Array.

```python
vector = torch.tensor([_ for _ in range(1, 6)])
print(vector)
print(vector.ndim)
```

#### iii) Matrix is a 2D array.

```python
MATRIX = torch.tensor([[_ for _ in range(1, 6)],
                       [_ for _ in range(6, 11)]])

print(MATRIX)
print(MATRIX.ndim)
```

#### Anything higher than 2 dimension is considered a tensor.

```python
TENSOR = torch.tensor([[[_ for _ in range(1, 6)],
                        [_ for _ in range(6, 11)]]])

print(TENSOR)
print(TENSOR.ndim)
```

## torch.rand()

#### While training a model, initially there will be random weights. This is implemented using rand() can be used. It gives the numbers in the range [0, 1].

#### Syntax: TENSOR = torch.rand(shape)

```python
TENSOR = torch.rand(1, 2, 3)

print(TENSOR)
print(TENSOR.ndim)
```

#### Tensors are the faster when compared to numpy arrays because tensors can natively perform parallel processing using GPU, TPU and other specialized hardwares. Numpy can do multi-core processing but it is limited to CPU.

## Tensor Attributes

#### i) shape
#### It shows the shape of the tensor.

```python
import torch

v = torch.tensor([[[1, 2, 3, 4],
                  [5, 6, 7, 8]]])
print(v.shape)
print(v.ndim)
```

## ii) dtype
#### It tells what type of data is stored in the tensor.
#### Common datatypes are int32, int64, bool, float32, float64

```python
import torch

v = torch.tensor([[[1, 2, 3, 4],
                  [5, 6, 7, 8]]])
print(v.dtype)
```

## iii) device
#### It tells where the tensor lives (GPU, CPU, TPU, etc)

```python
import torch

v = torch.tensor([[[1, 2, 3, 4],
                  [5, 6, 7, 8]]])
print(v.device)
```
