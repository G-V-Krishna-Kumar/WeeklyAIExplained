## Accessing Elements:
#### There are several ways to access specific elements or datapoints.
#### Assume the DataFrame is defined as below:

```python
import pandas as pd
import numpy as np

num = np.arange(1, 11)
pow = num**2

df = pd.DataFrame({"Number": num,
                  "Power": pow})
df.head()
```

### 1) df['col']
#### To access the whole column, df[<column_name>] is used.

```python
number_col = df['Number']
print(number_col)
print(type(number_col))
```

### 2) df.column_name
#### df.<column_name> can also be used instead of df['Number'] for retrieving a single column.

```python
number_col1 = df.Number
print(number_col1)
print(type(number_col1))
```

### 3) df[['col1', 'col2']]
#### To Access multiple columns, df[['col1', 'col2', ...]]

```python
number_and_power_col = df[['Number', 'Power']]
print(number_and_power_col)
print(type(number_and_power_col))
```

### 4) df.loc[row, col]
#### To access specific data by using the corresponding row and column.

```python
third_element = df.loc[2, 'Number']
third_element
```

### 5) df.loc[row]
#### To access entire row using the index label

```python
third_element = df.loc[2]
third_element
```

### 6) df.iloc[idx]
#### To access the entire row using the row number

```python
fourth_element = df.iloc[4]
fourth_element
```

### 7) df.iloc[row, col]
#### Access one specific value by numbers.

```python
fourth_element_power_column = df.iloc[3, 1]
fourth_element_power_column
```

### 8) df.at[row, col]
#### This gives the same output as df.loc[] but df.at[] is faster because it only takes the [row, col]. 
#### df.loc[] is like a general purpose accessor where loc has to determine if it is df.loc[row], df.loc[row, col], df.loc[condition], etc. This takes time.
#### Whereas, df.at[] only takes [row, col] so it can access the values faster

```python
fourth_element = df.at[3, 'Power']
fourth_element
```

#### To compare the execution times of loc[] and at[], we can use 'time' module in python.
#### Only for this example, new_df is created where columns are same. Only difference is there are 1M rows instead of 10 rows as in df. This is because, the execution time difference is noticeable only when the dataset is huge.

```python
import time 
import numpy as np

num = np.arange(1_000_000)
power = num ** 2

new_df = pd.DataFrame(
    {'num': num,
     'power': power}
)


# df,loc[]
start = time.perf_counter()
N = 1_000_000

for _ in range(N):
    x = new_df.loc[999999, "power"]

end = time.perf_counter()

print(f"loc: {end - start:.4f} seconds")



# df.at[]
start = time.perf_counter()

for _ in range(N):
    x = new_df.at[999999, "power"]

end = time.perf_counter()

print(f"at : {end - start:.4f} seconds")
```

#### at is 2.2x faster than loc in this case. Even though the execution times are different, df.at is almost always 2x faster than df.loc

### 9) df.iat[row, col_number]
#### i stands for integer. iat[] is used to access a specific data using the index values (or integer values) of the dataframe.

```python
fourth_element = df.iat[3, 1]
fourth_element
```

### 10) df.loc[start:end]
#### Splicing where index labels are used and the end is included

```python
rows = df.loc[2:6]
rows
```

### 11) df.iloc[start:end]
#### Splicing where the end is excluded. Integer position is used

```python
rows = df.iloc[2:6]
rows
```

### 12) df[condition]
#### Filter rows using 1 condition

```python
filtered = df[df["Number"] > 4]
filtered
```

#### The condition df[col] > 4 for example, will result in a Boolean values at each row. So, we return the rows where the Boolean value is True. This is called a Boolean Series or a Boolean Mask.

```python
number_col = df["Number"] >  4

print(type(number_col))
number_col
```

### df[(condition1) & (condition2)] 
#### Multiple conditions can be used using the logical operators

```python
filtered = df[(df["Number"]>4) & (df["Number"]<=8)]
filtered
```
