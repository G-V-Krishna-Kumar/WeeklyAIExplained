#### Assume the DataFrame is defined as below:

```python
import pandas as pd
import numpy as np

num_lst = [1, 2, 3, np.nan, 5, 6, np.nan]
num = np.array(num_lst)
pow = num**2

df = pd.DataFrame({"Number": num,
                  "Power": pow})
df
```

### 1) df.isna()/df.isnull()
#### Both are same and it returns a DataFrame with Boolean Values
#### It will assign True if the value is Null, else False.

```python
df.isna()
```

```python
df.isnull()
```

### 2) df.notna()/df.notnull()
#### Both are same and it returns a DataFrame with Boolean Values
#### It will assign False if the value is Null, else True.

#### It is the opposite of isna(), isnull()

```python
df
```

```python
df.notna()
```

```python
df.notnull()
```

### 3) df.fillna(value)
#### Fills the NAN (Not A Number) values with the specified value
#### It returns a DataFrame with the NAN values replaced with the specified value

```python
df.head()
```

```python
df = df.fillna(0)
df.head()
```
