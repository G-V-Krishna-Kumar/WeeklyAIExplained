#### Assume the DataFrame is defined as below:

```python
import pandas as pd
import numpy as np

num = np.arange(1, 11)
pow = num**2

df = pd.DataFrame({"Number": num,
                  "Power": pow})
df.head()
```

### 1) df["new_column"] = values
#### Adds a new column to the DataFrame

```python
df["Cube"] = df["Number"]**3
df.head()
```

### 2) df["column"] = values
#### Modifies existing column

```python
df["Power"] = df["Power"] * 10
df.head()
```

### 3) df.insert(loc, column, values)
#### Insert a new column at a specific position

```python
df.insert(1, "Fourth_Power", df["Number"]**4)

df.head()
```

### 4) df.drop(labels, axis=0/1)
#### Removes specified rows and columns and returns the new DataFrame
#### axis=0 means drop the specific row
#### axis=1 means drop the whole column

```python
df.head()
```

```python
df = df.drop("Fourth_Power", axis=1)
df.head()
```

### 5) del df[column_name]
#### Deletes the entire column
#### It does not return a DataFrame and it cannot delete rows

```python
df.head()
```

```python
del df["Cube"]
```

```python
df.head()
```

### 6) df.sort_values()
#### Sorts the DataFrame by column values

```python
df.sort_values("Power", ascending=False)
```
