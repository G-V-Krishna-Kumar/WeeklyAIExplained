# Eigenvalue and Eigenvector

#### Assume a transformation matrix A. If a vector x is given, performing Ax will result in a new vector. When there are a few vectors such that when the transformation is performed, it will scale or change to opposite direction. Such vectors are called eigenvectors and the scaling factor is called the eigenvalue.

#### **Eigenvector**: The vector such that after transformation, it will either stretch or doesn't change direction (The only direction it can change is flip in opposite direction. Basically, it just means scaling with negative number)

#### **Eigenvalue**: The scaled factor where it tells us how much the vector got stretched from the transformation.

#### The first equation tells that the performaing transformation on x is the same as scaling x with a scaling factor λ.
#### The goal is to find these set of eigenvectors (x) with their respective eigenvalue (λ). To do this, all the x terms are grouped so it becomes the equation in line 2: (A - λI)x = 0. 
#### We need to find the x such that the transformation A-λI is 0 and it must have non-zero x. In other words, (A-λI) must be singular. Singular matrices has determinant 0.
#### Hence, to calculate λ, we can use det(A-λI) = 0
